1. Write a function areatriangle(b,h) to compute the area of a triangle: formula is area = .5*b*h. Output should look like:
The area of a triangle of base 3 and height 5 is 7.5

```python
def areatriangle(b, h):
    """This function calculates the area of a triangle
    parameters: base(b) and height(h)"""
    area = 0.5 * b * h
    return area

# Call the function, store the returned value, and print the required statement
base = 3
height = 5
triangle_area = areatriangle(base, height)
print(f"The area of a triangle of base", base, "and height", height, "is", triangle_area)
```

```python
    def proteins_from_rf(self, aa_seq):
        """Compute all possible proteins in an aminoacid seq and return a list of possible proteins"""
        current_prot = []
        proteins = []
        for aa in aa_seq:
            if aa == "_":
                # STOP accumulating amino acids if _ - STOP was found
                if current_prot:
                    for p in current_prot:
                        proteins.append(p)
                    current_prot = []
            else:
                # START accumulating amino acids if M - START was found
                if aa == "M":
                    current_prot.append("")
                for i in range(len(current_prot)):
                    current_prot[i] += aa
        return proteins
```
